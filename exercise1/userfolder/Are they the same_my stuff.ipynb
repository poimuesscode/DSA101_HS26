```python
%matplotlib inline

import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from pathlib import Path
```

# Do we think the same thing is happening in all of these countries?

This is a hypothetical dataset representing pesticide residues above a safety threshold vs. percentage of organic farms in different countries. How can we tell if trends are consistent across countries?

```python
for file in Path('.').glob("*.csv"):
    df = pd.read_csv(file) #lädt in jedem Durchlauf die aktuelle Datei
    x, y = df['x'], df['y']
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    coeff = np.polyfit(x, y, deg=1)
    corr = np.corrcoef(x, y)[0, 1]
    
    print(f"x_mean={x_mean:.3f}, y_mean={y_mean:.3f}, linear regression fr(x)={coeff[1]:.1f} {coeff[0]:.2f} x, r = {corr:.3f} ")
    yline = coeff[1] + coeff[0] * x

    plt.figure() #startet eine neue, leere Grafik damit nicht alle Datenpunkte in die gleiche Abb. geplottet werden
    plt.scatter(df["x"], df["y"])
    plt.title(file.stem)
    plt.xlabel("organic farms (%)")
    plt.ylabel("pesticide residue above safety threshold (%)")
    plt.plot(x, yline, color="red")
    plt.show()
    
#the means, line regression, etc. (summery data) are very close across countries, suggesting a trend but visualization says clearly not. Scatter plots all have very different pattern
#the linear regression doesn't really fit the data, we can tell by looking at the r value (0.00 to ±0.19: Very weak or no linear relationship) 
# take home message: look at the data, statistics cannot be blindly trusted    
```
